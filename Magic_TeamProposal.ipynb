# Beyond the Earnings Surprise: Can AI Predict Post-Earnings Stock Returns?

By Team Magic: Abby Fischler, Gus Maass, Cammy Marks, and Michelle Orloff

## Research Question

For our final project, we want to use Python and AI to better understand how stock prices react to earnings announcements and whether AI analysis of earnings calls can help predict future stock performance.

While stock prices often react immediately to earnings announcements, research has shown that prices can continue moving in the direction of an earnings surprise for weeks afterward. This phenomenon is known as post-earnings announcement drift (PEAD).

We want to investigate whether this pattern still exists in recent financial markets and whether AI can identify additional information from earnings calls that traditional earnings metrics may overlook. For example, a company might beat earnings expectations while management expresses concerns about future demand or lowers guidance. We are interested in whether these qualitative signals help explain why some stocks continue outperforming after earnings while others reverse direction.

If we find that AI-generated sentiment scores improve predictions of future returns beyond traditional earnings surprise measures, our findings could be useful to investors, equity research analysts, and portfolio managers. More broadly, this project could demonstrate how AI can help investors process large amounts of financial information and identify patterns that are difficult to capture using numerical data alone.

### Research Question

Do stocks experience abnormal returns following earnings surprises, and can AI analysis of earnings calls improve predictions of subsequent stock performance?

### Our Hypothesis

We predict that companies reporting positive earnings surprises will, on average, experience positive abnormal returns in the weeks following their announcements, while companies reporting negative surprises will experience negative abnormal returns.

We also predict that AI-generated sentiment scores from earnings calls will provide additional predictive value. Specifically, companies with positive earnings surprises and optimistic management commentary may experience stronger subsequent returns than companies with similar earnings surprises but more cautious management commentary.

## Proposed Methodology

We plan to begin with a traditional post-earnings announcement drift analysis and expand to AI-based earnings call analysis if the necessary transcripts are available.

### Pseudocode

1. Gather a list of approximately 100 publicly traded companies.
2. Collect historical earnings announcement dates for each company.
3. Retrieve actual and expected earnings per share for each announcement.
4. Calculate the earnings surprise percentage.
5. Download historical daily stock returns.
6. Download market and factor returns.
7. Match earnings announcements to the appropriate trading dates, accounting for after-market releases.
8. Calculate announcement-period abnormal returns and subsequent 5-day, 20-day, and 60-day abnormal returns.
9. Sort earnings announcements into quintiles based on earnings surprise.
10. Compare average subsequent abnormal returns across quintiles.
11. Test whether earnings surprises are statistically associated with subsequent abnormal returns.
12. If transcripts are available, collect the corresponding earnings call text.
13. Use an AI model to generate sentiment, guidance, and uncertainty scores using a consistent rubric.
14. Merge the AI-generated scores with the earnings and stock return dataset.
15. Compare the predictive performance of models using earnings surprises alone against models incorporating AI-derived measures.
16. Evaluate results using chronological out-of-sample testing and account for transaction costs when considering potential trading strategies.
17. Export final datasets, tables, and figures.
18. Build an interactive dashboard presenting our results.

We will ensure that all information used for prediction was publicly available before the start of the return measurement period.

## Potential Challenges and Limitations

One challenge is ensuring that our analysis does not contain look-ahead bias. Earnings announcements released after market hours must be assigned to the appropriate trading period.

Another concern is survivorship bias. If we select companies based only on current index membership, our historical sample may exclude firms that were removed from the index.

For the AI extension, we must ensure that sentiment scores are based only on the original earnings call text, rather than information about stock performance that became available afterward.

We will also need to consider overlapping return periods, differences in firm size, transaction costs, and whether statistically significant results are economically meaningful.

Finally, transcript availability may limit the number of companies and years included in the AI portion of the analysis.
