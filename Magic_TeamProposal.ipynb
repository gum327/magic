# Beyond the Earnings Surprise: Do Earnings-Call Signals Improve Predictions of Post-Earnings Returns?

By Team Magic: Abby Fischler, Gus Maass, Cammy Marks, and Michelle Orloff

## Research Question

Our project asks whether pre-specified signals from management's earnings-call language improve chronological holdout prediction of post-earnings returns beyond conventional earnings data.

While stock prices often react immediately to earnings announcements, research has shown that prices can continue moving in the direction of an earnings surprise for weeks afterward. This phenomenon is known as post-earnings announcement drift (PEAD).

Post-earnings announcement drift (PEAD) is well established; rediscovering that positive surprises tend to precede positive returns is not our contribution. We will use a surprise-only model as a necessary baseline, then test whether contemporaneous commentary about outlook, guidance, and uncertainty adds predictive value.

Any AI extension is conditional on adequate transcript coverage and permission to use the text. The proposed pinned model is gpt-5.2-2025-12-11, published before the cohort and documented with an August 31, 2025 knowledge cutoff. We will report null or negative results and will not present prediction as proof of a profitable strategy.

### Research Question

For S&P 500 constituents as of June 30, 2026, do pre-specified signals from management's earnings-call language improve chronological holdout prediction of 20-trading-session factor-proxy-adjusted returns beyond EPS surprise and the announcement reaction for fiscal quarters ending in calendar Q2 2026?

### Our Hypothesis

EPS surprise and the announcement reaction provide a quantitative benchmark for subsequent abnormal returns; the project does not assume this established PEAD pattern is novel.

Among events with authorized, usable transcripts, we will test whether pre-specified call-language measures improve chronological holdout prediction over the baseline on the same events. This hypothesis is conditional on access and adequate coverage.

### What Are the Sample Conditions?

To be included in our dataset, a company must:

- Be publicly traded on a major U.S. stock exchange during the relevant period.
- Have identifiable quarterly earnings announcement dates.
- Have historical stock price data available around the earnings announcement.
- Have reported and expected earnings per share data available.
- Have sufficient subsequent trading data to calculate the selected return window.

For the AI extension, companies must also have accessible earnings call transcripts associated with their announcements.

This fixed, single-quarter cohort is not a multi-year sample selected from current index constituents, so the prior long-horizon survivorship criticism does not describe the target design. The dated roster prevents look-ahead; retain subsequent index departures.

## Proposed Methodology

The core test compares a quantitative earnings baseline with the same model plus pre-specified call-language measures, if lawful transcript access and coverage permit.

Measure the announcement reaction as the close-to-close return ending on the first tradable session. Start the forward 5-, 20-, and 60-session windows on the following session so the announcement-reaction predictor does not overlap the return outcome. The primary horizon is 20 sessions; 5 and 60 sessions are secondary when complete.

Estimate each firm's loadings on 252 pre-event daily returns ending before the first tradable session. Provisional ETF factors are market excess (SPY-BIL), size (IWM-SPY), and value (IWD-IWF). For each post-reaction day, calculate stock excess return minus the fitted intercept and factor components; sum daily residuals over the horizon. These are ETF proxies, not official French factors.

### Pseudocode

1. Validate and retain the archived June 30, 2026 constituent roster and document its non-official source.
2. Identify constituents with fiscal quarters ending in calendar Q2 2026; record fiscal period ends, releases, and call timestamps separately.
3. Audit actual/consensus EPS, timestamp, price, and ETF coverage; preserve exclusions and missingness.
4. Align each release to the first trading session when investors could act.
5. Calculate the announcement reaction through the first tradable session; begin the 5-, 20-, and 60-session return windows on the following session.
6. Estimate pre-event loadings on 252 daily observations using SPY-BIL, IWM-SPY, and IWD-IWF ETF proxies; aggregate residuals only after the announcement-reaction session. Do not label the proxies official French factors.
7. If transcript rights and broader cohort coverage are adequate, sort events by announcement timestamp, fit on the earlier 60%, and evaluate once on the later 40%. Use identical transcript-covered events for both models, and do not tune on the holdout.
8. Audit 10-20 calls for coverage only after confirming the source allows that check; record metadata only in inputs/transcript_access_audit_template.csv. This is a feasibility check, not a sufficient prediction sample. Obtain explicit text-retention and external-model permission before collecting or scoring any transcript.
9. Score authorized call text with the pinned gpt-5.2-2025-12-11 snapshot and a fixed rubric; record the provider's documented cutoff and terms. Do not use a floating model alias.
10. Compare mean absolute error, root mean squared error, and out-of-sample R-squared relative to the training-set outcome mean on identical events. Report uncertainty and coverage changes; if the usable holdout is too small, label the analysis exploratory or defer it.
11. Export reproducible tables and figures and document limitations; do not imply profitability without a separate costed strategy test.

Use only predictor information available by the first tradable session. Keep later returns and post-call information out of features and AI scores.

## Potential Challenges and Limitations

One challenge is ensuring that our analysis does not contain look-ahead bias. Earnings announcements released after market hours must be assigned to the appropriate trading period.

The fixed June 30, 2026 roster avoids selecting the one-quarter cohort from current membership; keep firms that leave the index afterward. This is not the former multi-year current-constituent design.

For the AI extension, verify reuse rights and a pre-cohort training-data cutoff. Scores must use only original call text, never later prices or post-call information.

We will also need to consider overlapping return periods, differences in firm size, transaction costs, and whether statistically significant results are economically meaningful.

Finally, transcript access and event alignment may limit the AI sample; public Seeking Alpha pages do not establish reuse rights, and automated scraping is prohibited by its current terms.

## Expected Outcome

We will estimate whether authorized, pre-specified earnings-call signals improve chronological holdout predictions beyond EPS surprise and announcement reaction for this fixed Q2 2026 cohort. The 10-20-call audit is only a feasibility check; the cohort or holdout may be too small for stable inference, and null results are possible.

Our final dashboard will allow users to explore these relationships and better understand how financial markets respond to both quantitative earnings results and qualitative management commentary.

The current Yahoo/SPY four-company pilot is a pipeline check, not a result for the target cohort. See README.md, PROPOSAL.md, and STATUS.md for the current plan and access status.
